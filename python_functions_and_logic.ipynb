```python
# @title
def evaluate_function(func, test_cases):
    """
    Chạy các test case để đánh giá một hàm cụ thể.
    """
    results = {"passed": 0, "failed": 0, "details": []}

    for i, test_case in enumerate(test_cases, 1):
        try:
            if "kwargs" in test_case:
                result = func(*test_case["input"], **test_case["kwargs"])
            else:
                result = func(*test_case["input"])
            expected = test_case["expected"]
            passed = result == expected

            if passed:
                results["passed"] += 1
            else:
                results["failed"] += 1
                results["details"].append(
                    f"Test case {i} failed: Input={test_case['input']}, "
                    f"Expected={test_case['expected']}, Got={result}"
                )
        except Exception as e:
            if isinstance(test_case["expected"], type) and isinstance(e, test_case["expected"]):
                results["passed"] += 1
            else:
                results["failed"] += 1
                results["details"].append(
                    f"Test case {i} error: Input={test_case['input']}, Error={str(e)}"
                )

    print(f"Passed: {results['passed']}")
    print(f"Failed: {results['failed']}")
    if results["failed"] > 0:
        print("Details of failed test cases:")
        for detail in results["details"]:
            print(detail)


test_cases_cau1 = [
    {"input": (5, 10), "expected": 57000},
    {"input": (0, 0), "expected": 12000},
    {"input": (2.5, 8), "expected": 36000},
    {"input": (10, 0), "expected": 92000},
    {"input": (1, 0.0001), "expected": 20000},
]

test_cases_cau2 = [
    {"input": (1350000,), "expected": (2, 3, 1)},
    {"input": (0,), "expected": (0, 0, 0)},
    {"input": (50000,), "expected": (0, 0, 1)},
    {"input": (450000,), "expected": (0, 4, 1)},
    {"input": (650000,), "expected": (1, 1, 1)},
    {"input": (2000000,), "expected": (4, 0, 0)},
]

test_cases_cau3 = [
    {"input": (0,), "expected": 0},
    {"input": (8000000,), "expected": 400000},
    {"input": (10000000,), "expected": 500000},
    {"input": (20000000,), "expected": 1500000},
    {"input": (30000000,), "expected": 2500000},
    {"input": (40000000,), "expected": 4500000},
    {"input": (50000000,), "expected": 6500000},
]

test_cases_cau4 = [
    {"input": (0, 0, 0, 0, 10), "expected": "Nam trong hinh vuong"},
    {"input": (3, -4, 0, 0, 10), "expected": "Nam trong hinh vuong"},
    {"input": (5, 0, 0, 0, 10), "expected": "Nam tren canh"},
    {"input": (5, 5, 0, 0, 10), "expected": "Nam tren canh"},
    {"input": (3, 1, 2, 2, 2), "expected": "Nam tren canh"},
    {"input": (6, 0, 0, 0, 10), "expected": "Nam ngoai hinh vuong"},
    {"input": (3.5, 2, 2, 2, 2), "expected": "Nam ngoai hinh vuong"},
    {"input": (-1, 7, 0, 0, 4), "expected": "Nam ngoai hinh vuong"},
]

test_cases_cau5 = [
    {"input": (100, 10), "expected": (8, 214.36)},
    {"input": (1000, 100), "expected": (1, 2000.0)},
    {"input": (500, 50), "expected": (2, 1125.0)},
    {"input": (200, 25), "expected": (4, 488.28)},
]

test_cases_cau6 = [
    {"input": (160,), "expected": 4000000},
    {"input": (160, 30000), "expected": 4800000},
    {"input": (160,), "kwargs": {"thuong": 500000}, "expected": 4500000},
    {"input": (100,), "kwargs": {"don_gia": 40000, "thuong": 200000}, "expected": 4200000},
    {"input": (0,), "expected": 0},
    {"input": (120.5, 20000), "expected": 2410000.0},
]

test_cases_cau7 = [
    {"input": (3, 8, 5), "expected": (3, 8)},
    {"input": (-5, -2, -9), "expected": (-9, -2)},
    {"input": (7,), "expected": (7, 7)},
    {"input": (2.5, 2.5, 1), "expected": (1, 2.5)},
    {"input": (), "expected": ValueError},
]

test_cases_cau8 = [
    {"input": (1, 2, 4), "expected": 2.33},
    {"input": (5,), "expected": 5.0},
    {"input": (10, 20), "expected": 15.0},
    {"input": (-3, 3), "expected": 0.0},
    {"input": (1.5, 2.5, 3.5), "expected": 2.5},
    {"input": (), "expected": ValueError},
]

test_cases_cau9 = [
    {"input": ([["Toan", 10], ["Van", 25], ["Anh", 25]],), "expected": (60, "Van")},
    {"input": ([["A", 5]],), "expected": (5, "A")},
    {"input": ([["A", 0], ["B", 0]],), "expected": (0, "A")},
    {"input": ([["Python", 120], ["Java", 80], ["C++", 95], ["Go", 120]],), "expected": (415, "Python")},
]

test_cases_cau10 = [
    {"input": ([9, 8, 7, 6.5, 5, 4],), "expected": [2, 2, 1, 1]},
    {"input": ([],), "expected": [0, 0, 0, 0]},
    {"input": ([10, 10, 8],), "expected": [3, 0, 0, 0]},
    {"input": ([4.9, 0],), "expected": [0, 0, 0, 2]},
    {"input": ([7.9, 6.49, 5.0, 7.99, 8.0],), "expected": [1, 2, 2, 0]},
]
```

## Question 2: Currency Bill Breakdown
Write a function `exchange_money(amount)` that takes a non-negative integer `amount` (always a multiple of `50000`) and returns a tuple `(bills_500k, bills_100k, bills_50k)` representing the number of `500000 VND`, `100000 VND`, and `50000 VND` bills such that the **total number of bills is minimized** (prioritizing higher denominations first).

**Hint**: Use integer division `//` and modulo `%`.


```python
def exchange_money(amount):
    """
    Break down an amount of money into the minimum number of 500k, 100k, and 50k bills.

    Args:
    - amount (int): Non-negative amount (multiple of 50000).

    Returns:
    - tuple: (bills_500k, bills_100k, bills_50k)
    """
    bills_500k = amount // 500000
    remainder = amount % 500000
    
    bills_100k = remainder // 100000
    remainder = remainder % 100000
    
    bills_50k = remainder // 50000
    
    return (bills_500k, bills_100k, bills_50k)

# Test the function
evaluate_function(exchange_money, test_cases_cau2)
```

Write a function `calculate_income_tax(income)` to calculate progressive income tax (in VND):

- Income portion from `0` to `10 000 000`: `5%` tax.
- Income portion above `10 000 000` up to `30 000 000`: `10%` tax.
- Income portion above `30 000 000`: `20%` tax.

**Note**: each tax rate applies only to the portion of income within that bracket (do not apply the highest tax rate to the entire income).


```python
def calculate_income_tax(income):
    """
    Calculate progressive income tax based on tax brackets.

    Args:
    - income (float): Income (non-negative).

    Returns:
    - float: Total income tax.

    Examples:
    >>> calculate_income_tax(8000000)
    400000.0
    >>> calculate_income_tax(40000000)
    4500000.0
    """
    tax = 0
    if 0 <= income <= 10000000:
        tax = income * (5 / 100)
    else:
        if 10000000 < income <= 30000000:
            tax = 10000000 * (5 / 100) + (income - 10000000) * (10 / 100)
        else:
            if income > 30000000:
                tax = 10000000 * (5 / 100) + 20000000 * (10 / 100) + (income - 10000000 - 20000000) * (20 / 100)
    return tax

# Test the function
evaluate_function(calculate_income_tax, test_cases_cau3)
```

Write a function `point_square_position(x, y, cx, cy, a)` to check the position of a point `(x, y)` relative to a square centered at `(cx, cy)` with side length `a` and sides parallel to the coordinate axes. The function returns one of three strings:
- `"Inside the square"` if the point lies strictly inside.
- `"On the edge"` if the point lies on the boundary (including vertices).
- `"Outside the square"` if the point lies outside.


```python
def point_square_position(x, y, cx, cy, a):
    """
    Check the position of a point relative to a square (sides parallel to the axes).

    Args:
    - x, y (float): Coordinates of the point to check.
    - cx, cy (float): Coordinates of the square's center.
    - a (float): Side length of the square (a > 0).

    Returns:
    - str: "Nam trong hinh vuong" / "Nam tren canh" / "Nam ngoai hinh vuong".

    Examples:
    >>> point_square_position(0, 0, 0, 0, 10)
    'Nam trong hinh vuong'
    """
    half = a / 2
    dx = abs(cx - x)
    dy = abs(cy - y)
    
    if dx < half and dy < half:
        return "Nam trong hinh vuong"
    else:
        if dx <= half and dy <= half:
            return "Nam tren canh"
        else:
            return "Nam ngoai hinh vuong"

# Test the function
evaluate_function(point_square_position, test_cases_cau4)
```

## Question 5: Calculate Years to Double Principal
Write a function `years_to_double(P, r)` to calculate the minimum number of years required for an initial savings deposit `P` with an annual compound interest rate of `r`% to reach or exceed **double** the initial amount (`2 * P`). The function returns two values `(years, amount)` with `amount` rounded to 2 decimal places.

Yearly formula: $P_{new} = P_{old} \times (1 + \frac{r}{100})$

(Given `P > 0`, `r > 0`.)

**Hint**: Use a `while` loop.



```python
def years_to_double(P, r):
    """
    Calculate the number of years required for the deposit (compound interest) to double.

    Args:
    - P (float): Initial deposit amount (P > 0).
    - r (float): Annual interest rate (percentage, r > 0).

    Returns:
    - (years, amount): required number of years and final amount (rounded to 2 decimal places).

    Examples:
    >>> years_to_double(100, 10)
    (8, 214.36)
    """
    years = 0
    amount = P
    while amount < 2 * P:
        amount = amount * (1 + (r / 100))
        years = years + 1
    return years, round(amount, 2)

# Test the function
evaluate_function(years_to_double, test_cases_cau5)
```

## Question 7: Find Minimum and Maximum Numbers
Write a function `find_min_max(*so)` that takes an arbitrary number of numeric arguments and returns a tuple of two values `(smallest, largest)`. If no arguments are passed, raise a `ValueError`.

**Note:** Do not use the built-in `min()` and `max()` functions.



```python
def find_min_max(*so):
    """
    Find the smallest and largest numbers from the given arguments.

    Args:
    - *so: Numbers (arbitrary count).

    Returns:
    - (smallest, largest).

    Raises:
    - ValueError: If no numbers are provided.

    Examples:
    >>> find_min_max(3, 8, 5)
    (3, 8)
    """
    if len(so) == 0:
        raise ValueError("Phai truyen it nhat mot so.")

    smallest = so[0]
    largest = so[0]
    for x in so:
        if x < smallest:
            smallest = x
        if x > largest:
            largest = x
    return smallest, largest

# Test the function
evaluate_function(find_min_max, test_cases_cau7)
```

## Question 8: Calculate Arithmetic Mean
Write a function `calculate_average(*so)` that takes an arbitrary number of numeric arguments and returns their arithmetic mean, rounded to 2 decimal places. If no arguments are passed, raise a `ValueError`.

**Note:** Do not use the built-in `sum()` function.


```python
def calculate_average(*so):
    """
    Calculate the arithmetic mean of the given numbers.

    Args:
    - *so: Numbers (arbitrary count).

    Returns:
    - float: Arithmetic mean, rounded to 2 decimal places.

    Raises:
    - ValueError: If no numbers are provided.

    Examples:
    >>> calculate_average(1, 2, 4)
    2.33
    """
    if len(so) == 0:
        raise ValueError("Phai truyen it nhat mot so.")

    total = 0
    for x in so:
        total += x
    return round(total / len(so), 2)

# Test the function
evaluate_function(calculate_average, test_cases_cau8)
```

## Question 9: Analyze Library Book Borrows
Write a function `analyze_library(ds)` that takes a list of books (each book is `[book_title, borrow_count]`) and returns two values `(total_borrows, most_borrowed_book)`, where:
- `total_borrows` is the sum of borrow counts across all books.
- `most_borrowed_book` is the title of the book with the highest `borrow_count`. If there is a tie, return the book that appears **first** in the list.

(The list always contains at least 1 book.)


```python
def analyze_library(ds):
    """
    Analyze library book borrow counts to find total borrows and the most borrowed book.

    Args:
    - ds (list): List of [book_title, borrow_count] pairs (non-empty).

    Returns:
    - tuple: (total_borrows, most_borrowed_book)

    Examples:
    >>> analyze_library([["Python Basics", 10], ["Data Science", 15]])
    (25, 'Data Science')
    """
    total_borrows = 0
    most_borrowed_book = ds[0][0]
    max_borrows = ds[0][1]

    for book in ds:
        title = book[0]
        count = book[1]
        
        total_borrows += count
        if count > max_borrows:
            max_borrows = count
            most_borrowed_book = title

    return total_borrows, most_borrowed_book

# Test the function
evaluate_function(analyze_library, test_cases_cau9)
```

## Question 10: Academic Performance Classification Statistics
Write a function `classify_grades(ds_diem)` that takes a list of students' average scores (`ds_diem` is a list of floats, which may be empty) and returns a **list of 4 integers** `[excellent_count, good_count, average_count, weak_count]` counting the number of students in each academic performance category:
- Excellent: score `>= 8`.
- Good: `6.5 <= score < 8`.
- Average: `5 <= score < 6.5`.
- Weak: score `< 5`.

The returned list must always contain all 4 elements in the exact order above (if a category has no students, its count is `0`).


```python
def classify_grades(ds_diem):
    """
    Classify student scores into each academic performance category.

    Args:
    - ds_diem (list): List of students' average scores.

    Returns:
    - list: [excellent_count, good_count, average_count, weak_count].

    Examples:
    >>> classify_grades([9, 7, 4])
    [1, 1, 0, 1]
    """
    stats = [0, 0, 0, 0]
    for score in ds_diem:
        if score >= 8:
            stats[0] += 1
        else:
            if score >= 6.5:
                stats[1] += 1
            else:
                if score >= 5:
                    stats[2] += 1
                else:
                    stats[3] += 1

    return stats

# Test the function
evaluate_function(classify_grades, test_cases_cau10)
```
